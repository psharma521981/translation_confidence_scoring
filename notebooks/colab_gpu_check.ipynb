```python
# Quick GPU speed test (small matrix multiply)
import time
if torch.cuda.is_available():
    x = torch.randn(4096, 4096, device="cuda")
    torch.cuda.synchronize(); t = time.time()
    for _ in range(20):
        y = x @ x
    torch.cuda.synchronize()
    print(f"20 matmuls took {time.time() - t:.2f}s")
else:
    print("No GPU - skipping speed test")
```
